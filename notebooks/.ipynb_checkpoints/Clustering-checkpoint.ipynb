```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt

cities = pd.read_csv("../CitiesDB_with_polycentrism.csv")
zib = pd.read_csv("../zib_fitting_results.csv")
height = pd.read_csv("../height_fitting_results(log_ols).csv")

# Unify column names
zib = zib.rename(columns={'city': 'Name'})
height = height.rename(columns={'city': 'Name'})

# Merge the two model result tables
merged = zib.merge(height, on=["Name", "year"], how="inner")

# Pivot to wide format (2016 and 2023 in columns)
wide = merged.pivot(index="Name", columns="year")
wide.columns = [f"{param}_{year}" for param, year in wide.columns]
wide = wide.reset_index()

# Merge with city metadata
df = cities.merge(wide, on="Name", how="inner")
feature_cols = [
    "alpha_zi_2016", "beta_zi_2016", "alpha_bin_2016", "beta_bin_2016",
    "A_2016", "B_2016", "C_2016"
    # "alpha_zi_2023", "beta_zi_2023", "alpha_bin_2023", "beta_bin_2023",
    # "A_2023", "B_2023", "C_2023"
]
 
# for c in [ "alpha_zi", "beta_zi","alpha_bin", "beta_bin", "A", "B", "C"]:
#     df[f"{c}_change"] = df[f"{c}_2023"] - df[f"{c}_2016"]
#     feature_cols.append(f"{c}_change")

print(feature_cols)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[feature_cols])
```

```python
# ========== best k ==========
scores = []
for k in range(2, 10):
    km = KMeans(n_clusters=k, random_state=42)
    labels = km.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    scores.append((k, score))

best_k = max(scores, key=lambda x: x[1])[0]
print("best k =", best_k)

plt.plot([k for k, _ in scores], [s for _, s in scores], marker='o')
plt.xlabel("k")
plt.ylabel("Silhouette Score")
plt.title("choose best k")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# --- Run clustering ---
k = 4
kmeans = KMeans(n_clusters=k, random_state=42)
df["cluster"] = kmeans.fit_predict(X_scaled)

# --- Define a consistent color palette ---
palette = sns.color_palette("tab10", n_colors=k)
cluster_colors = {i: palette[i] for i in range(k)}  # dict for easy lookup

# --- PCA Visualization ---
pca = PCA(n_components=2)
X_vis = pca.fit_transform(X_scaled)

plt.figure(figsize=(9,7))
for i in range(k):
    subset = df[df["cluster"] == i]
    plt.scatter(
        X_vis[subset.index, 0],
        X_vis[subset.index, 1],
        color=cluster_colors[i],
        label=f"Cluster {i}",
        s=60,
        alpha=0.8
    )
plt.xlabel("PCA 1 (for visualization only)")
plt.ylabel("PCA 2")
plt.title(f"Urban Morphology Clustering (K={k})", fontsize=13)
plt.grid(alpha=0.3)
for i, name in enumerate(df["Name"]):
    plt.text(X_vis[i,0], X_vis[i,1], name, fontsize=6, alpha=0.6)
plt.legend(title="Cluster")
plt.tight_layout()
plt.show()

# --- Cluster means ---
cluster_summary = df.groupby("cluster")[feature_cols].mean().round(3)
print("\n=== Mean parameters per cluster ===")
print(cluster_summary)

# --- Radar chart ---
# selected_features = [
#     "alpha_zi_change", "beta_zi_change",
#     "alpha_bin_change", "beta_bin_change",
#     "A_change", "B_change", "C_change"
# ]

selected_features = feature_cols
cluster_means = cluster_summary[selected_features].copy()
num_vars = len(selected_features)

angles = np.linspace(0, 2*np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]

plt.figure(figsize=(8,8), dpi=120)
for i in range(k):
    values = cluster_means.iloc[i].tolist()
    values += values[:1]
    plt.polar(angles, values, label=f'Cluster {i}', color=cluster_colors[i], linewidth=2)
plt.xticks(angles[:-1], selected_features, fontsize=8)
plt.title("Average Δ-Parameter Patterns per Cluster (Radar Chart)", size=13)
plt.legend(loc='upper right', bbox_to_anchor=(1.2, 1.1))
plt.tight_layout()
plt.show()
```

```python
# --- Feature columns for 2016 and 2023 ---
features_2016 = ['alpha_zi_2016','beta_zi_2016','alpha_bin_2016','beta_bin_2016','A_2016','B_2016','C_2016']
features_2023 = ['alpha_zi_2023','beta_zi_2023','alpha_bin_2023','beta_bin_2023','A_2023','B_2023','C_2023']

# --- Standardize separately (or together for consistency) ---
scaler = StandardScaler()
X_2016_scaled = scaler.fit_transform(df[features_2016])
X_2023_scaled = scaler.fit_transform(df[features_2023])  # 用同一scaler

# --- KMeans clustering on 2016 data ---
k = 4
kmeans_2016 = KMeans(n_clusters=k, random_state=42)
df['cluster_2016'] = kmeans_2016.fit_predict(X_2016_scaled)

# --- Assign 2023 data to 2016 clusters ---
from sklearn.metrics import pairwise_distances_argmin
df['cluster_2023_based_on_2016'] = pairwise_distances_argmin(X_2023_scaled, kmeans_2016.cluster_centers_)

# --- PCA for visualization (fit on 2016, project 2023) ---
pca = PCA(n_components=2)
X_2016_2D = pca.fit_transform(X_2016_scaled)
X_2023_2D = pca.transform(X_2023_scaled)

# --- Color palette ---
import seaborn as sns
palette = sns.color_palette("tab10", n_colors=k)
cluster_colors = {i: palette[i] for i in range(k)}

# --- PCA scatter with arrows ---
plt.figure(figsize=(10,8))
for i, name in enumerate(df['Name']):
    plt.arrow(
        X_2016_2D[i,0], X_2016_2D[i,1],
        X_2023_2D[i,0]-X_2016_2D[i,0],
        X_2023_2D[i,1]-X_2016_2D[i,1],
        head_width=0.05, alpha=0.3, color='gray'
    )

for c in range(k):
    idx = df['cluster_2016']==c
    plt.scatter(X_2016_2D[idx,0], X_2016_2D[idx,1], color=cluster_colors[c], label=f'2016 Cluster {c}', s=60)
    
    idx2 = df['cluster_2023_based_on_2016']==c
    plt.scatter(X_2023_2D[idx2,0], X_2023_2D[idx2,1], color=cluster_colors[c], marker='x', s=80, label=f'2023 in Cluster {c}')

plt.xlabel("PCA 1 (2016 projection)")
plt.ylabel("PCA 2")
plt.title("City Cluster Changes: 2016 → 2023")
plt.legend(bbox_to_anchor=(1.05,1), loc="upper left")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# --- Radar chart: compare cluster averages 2016 vs 2023 ---
cluster_summary_2016 = pd.DataFrame(kmeans_2016.cluster_centers_, columns=features_2016)
cluster_summary_2023 = pd.DataFrame(
    [X_2023_scaled[df['cluster_2023_based_on_2016']==i].mean(axis=0) for i in range(k)],
    columns=features_2016
)  # 对应2016标准化列名，方便对比

angles = np.linspace(0, 2*np.pi, len(features_2016), endpoint=False).tolist()
angles += angles[:1]

plt.figure(figsize=(8,8), dpi=120)
for i in range(k):
    values_2016 = cluster_summary_2016.iloc[i].tolist()
    values_2016 += values_2016[:1]
    values_2023 = cluster_summary_2023.iloc[i].tolist()
    values_2023 += values_2023[:1]
    plt.polar(angles, values_2016, label=f'Cluster {i} 2016', color=cluster_colors[i], linewidth=2)
    plt.polar(angles, values_2023, label=f'Cluster {i} 2023', color=cluster_colors[i], linewidth=2, linestyle='--')

plt.xticks(angles[:-1], features_2016, fontsize=8)
plt.title("Cluster Average Parameters: 2016 vs 2023", fontsize=13)
plt.legend(loc='upper right', bbox_to_anchor=(1.3,1.1))
plt.tight_layout()
plt.show()
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt
import numpy as np

# Compute hierarchical linkage across all individual cities
Z_full = linkage(X_scaled, method='ward')

# Optionally, order leaves by their assigned cluster
# (so cities from same cluster appear near each other)
# To do that, sort df and X_scaled by cluster:
sort_idx = np.argsort(df["cluster"])
df_sorted = df.iloc[sort_idx]
Z_full = linkage(X_scaled[sort_idx], method='ward')

# Plot dendrogram
plt.figure(figsize=(12, 6))
dendrogram(
    Z_full,
    labels=df_sorted["Name"].values,   # label each leaf with city name
    leaf_rotation=90,                  # rotate labels for readability
    leaf_font_size=7,                  # small font for many cities
    color_threshold=0,                 # color threshold disabled (we’ll color by cluster below)
)

plt.title("Hierarchical Clustering of All Cities (Each City = Leaf)", fontsize=13)
plt.xlabel("City")
plt.ylabel("Distance")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define your feature pairs (x,y)
pairs = [
    ("alpha_zi_2016", "beta_zi_2016"),
    ("alpha_bin_2016", "beta_bin_2016"),
    ("alpha_zi_2016", "A_2016"),
    ("A_2016", "B_2016"),
    ("A_2016", "C_2016"),
    ("alpha_zi_change", "beta_zi_change"),
    ("alpha_bin_change", "beta_bin_change"),
    ("A_change", "B_change"),
    ("A_change", "C_change")
]

# Create subplots
ncols = 3
nrows = int(np.ceil(len(pairs) / ncols))
fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(14,12))
axes = axes.flatten()

for i, (xvar, yvar) in enumerate(pairs):
    ax = axes[i]
    sns.scatterplot(
        data=df,
        x=xvar,
        y=yvar,
        hue="cluster",
        palette="tab10",
        s=50,
        alpha=0.8,
        ax=ax
    )
    ax.set_xlabel(xvar, fontsize=9)
    ax.set_ylabel(yvar, fontsize=9)
    ax.set_title(f"{xvar} vs. {yvar}", fontsize=10)
    ax.grid(alpha=0.3)

# Remove unused subplots if any
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

# Single legend on the right
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, title="Cluster", bbox_to_anchor=(1.02, 0.95), loc="upper left")

plt.suptitle("Clustered City Scatter Plots of α–β Relationships", fontsize=14)
plt.tight_layout(rect=[0, 0, 0.95, 0.95])
plt.show()
```

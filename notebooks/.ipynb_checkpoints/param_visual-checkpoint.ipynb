```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ===========================
# 读表
# ===========================
cities = pd.read_csv('../CitiesDB_new.csv')
zib = pd.read_csv('../zib_fitting_results_new.csv')

# ===========================
# 选用关键列
# ===========================
zib_sel = zib[['city','year','alpha','beta','kappa','delta','population']]

# 选城市信息
cities_sel = cities[['Name','continent','pop_2016_UN','pop_2023_UN']].copy()
cities_sel = cities_sel.rename(columns={'Name':'city'})

# ===========================
# 合并
# ===========================
df = pd.merge(zib_sel, cities_sel, on='city', how='left')
# 根据年份选择人口大小
def get_pop(row):
    if row['year']==2016:
        return row['pop_2016_UN']
    else:
        return row['pop_2023_UN']

df['pop_size'] = df.apply(get_pop, axis=1)

# 自定义颜色映射
color_map = {
    'Africa': '#E85748',
    'Asia': '#F39F1B',
    'Latin America': '#47A2DE'
}

# 绘图函数，带 legend
def scatter_plot(x, y, title, xlabel, ylabel):
    fig, ax = plt.subplots(figsize=(8,6))
    
    for year, alpha_val in [(2016, 0.4), (2023, 0.9)]:
        sub = df[df['year']==year]
        # 逐个大洲绘制，以便 legend
        for cont, color in color_map.items():
            sub_cont = sub[sub['continent']==cont]
            ax.scatter(
                sub_cont[x],
                sub_cont[y],
                s=np.sqrt(sub_cont['pop_size'])*1.5,
                c=color,
                alpha=alpha_val,
                edgecolor='k' if year==2023 else 'none',
                linewidth=0.5,
                label=f'{cont} {year}'
            )
    
    ax.set_xlabel(xlabel,fontsize=16)
    ax.set_ylabel(ylabel,fontsize=16)
    ax.set_title(title,fontsize=22)
    
    # 避免重复图例
    handles, labels = ax.get_legend_handles_labels()
    by_label = dict(zip(labels, handles))
    ax.legend(by_label.values(), by_label.keys(), title='Continent / Year', loc='best')
    
    plt.tight_layout()
    plt.show()

# α–β 散点图
scatter_plot('alpha', 'beta', 'Structural parameters', r'$\alpha$', r'$\beta$')

# κ–γ 散点图
scatter_plot('kappa', 'delta', 'Construct parameters', r'$\kappa$', r'$\delta$')
```

```python
from scipy.special import expit
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# α–β 图
# -----------------------------
fig, ax = plt.subplots(figsize=(6,6))

# 背景散点
for year, alpha_val in [(2016, 0.4), (2023, 0.9)]:
    sub = df[df['year']==year].copy()
    sub['p_alpha'] = expit(sub['alpha'])
    ax.scatter(sub['p_alpha'], sub['beta'],
               s=20,
               c='lightgray' if year==2016 else 'dimgray',
               alpha=alpha_val)

# 洲平均点
avg_points = df.groupby(['continent','year'])[['alpha','beta']].mean().reset_index()
for cont, color in color_map.items():
    for year, alpha_val in [(2016,0.6),(2023,0.9)]:
        sub = avg_points[(avg_points['continent']==cont) & (avg_points['year']==year)].copy()
        sub['p_alpha'] = expit(sub['alpha'])
        for _, row in sub.iterrows():
            ax.scatter(row['p_alpha'], row['beta'],
                       s=200,
                       c=color,
                       alpha=alpha_val,
                       edgecolor='k',
                       label=cont)  # 用洲名作为label

# 参考线：直接在坐标轴显示
ax.axvline(0.5, color='k', linestyle='--')
ax.text(0.5, ax.get_ylim()[1]*0.95, 'p=0.5', ha='center', va='top', fontsize=12)

mean_beta = df['beta'].mean()
ax.axhline(mean_beta, color='k', linestyle=':')
ax.text(ax.get_xlim()[1]*0.98, mean_beta, f'β={mean_beta:.2f}', ha='right', va='center', fontsize=12)

# 横轴设置
ax.set_xlim(0,1)
ax.set_xticks(np.arange(0,1.01,0.1))

ax.set_xlabel('α probability (p)', fontsize=16)
ax.set_ylabel(r'$\beta$', fontsize=16)
ax.set_title('Structural parameters α–β', fontsize=22)

# legend 去重，只显示洲名
handles, labels = ax.get_legend_handles_labels()
by_label = dict(zip(labels, handles))
ax.legend(by_label.values(), by_label.keys(), loc='best')
plt.tight_layout()
plt.show()

# -----------------------------
# κ–δ 图
# -----------------------------
fig, ax = plt.subplots(figsize=(6,6))

# 背景散点
for year, alpha_val in [(2016, 0.4), (2023, 0.9)]:
    sub = df[df['year']==year].copy()
    sub['p_kappa'] = expit(sub['kappa'])
    ax.scatter(sub['p_kappa'], sub['delta'],
               s=20,
               c='lightgray' if year==2016 else 'dimgray',
               alpha=alpha_val)

# 洲平均点
avg_points = df.groupby(['continent','year'])[['kappa','delta']].mean().reset_index()
for cont, color in color_map.items():
    for year, alpha_val in [(2016,0.6),(2023,0.9)]:
        sub = avg_points[(avg_points['continent']==cont) & (avg_points['year']==year)].copy()
        sub['p_kappa'] = expit(sub['kappa'])
        for _, row in sub.iterrows():
            ax.scatter(row['p_kappa'], row['delta'],
                       s=200,
                       c=color,
                       alpha=alpha_val,
                       edgecolor='k',
                       label=cont)

# 参考线
ax.axvline(0.5, color='k', linestyle='--')
ax.text(0.5, ax.get_ylim()[1]*0.95, 'p=0.5', ha='center', va='top', fontsize=12)

mean_delta = df['delta'].mean()
ax.axhline(mean_delta, color='k', linestyle=':')
ax.text(ax.get_xlim()[1]*1.2, mean_delta, f'δ={mean_delta:.2f}', ha='right', va='center', fontsize=12)

# 横轴设置
ax.set_xlim(0,1)
ax.set_xticks(np.arange(0,1.01,0.1))

ax.set_xlabel('κ probability (p)', fontsize=16)
ax.set_ylabel(r'$\delta$', fontsize=16)
ax.set_title('Construct parameters κ–δ', fontsize=22)

# legend 去重，只显示洲名
handles, labels = ax.get_legend_handles_labels()
by_label = dict(zip(labels, handles))
ax.legend(by_label.values(), by_label.keys(), loc='best')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 选择必要列
height_sel = height[['city','year','alpha','beta','gamma']].copy()
height_sel = height_sel.rename(columns={'alpha':'A','beta':'B','gamma':'C'})
cities_sel = cities[['Name','continent','pop_2016_UN','pop_2023_UN']].copy()
cities_sel = cities_sel.rename(columns={'Name':'city'})
height_sel = pd.merge(height_sel, cities_sel[['city','continent','pop_2016_UN','pop_2023_UN']], on='city', how='left')

# 添加人口大小
height_sel['pop_size'] = height_sel.apply(lambda row: row['pop_2016_UN'] if row['year']==2016 else row['pop_2023_UN'], axis=1)

# 三联图
fig, axes = plt.subplots(1,3,figsize=(18,6))
pairs = [('A','B'), ('B','C'), ('A','C')]
titles = ['A–B', 'B–C', 'A–C']

for ax, (x,y), title in zip(axes, pairs, titles):
    # 背景散点
    for year, alpha_val in [(2016,0.4),(2023,0.9)]:
        sub = height_sel[height_sel['year']==year]
        ax.scatter(sub[x], sub[y],
                   s=20,
                   c='lightgray' if year==2016 else 'dimgray',
                   alpha=alpha_val)
    
    # 洲平均点
    avg_points = height_sel.groupby(['continent','year'])[[x,y]].mean().reset_index()
    for cont, color in color_map.items():
        for year, alpha_val in [(2016,0.6),(2023,0.9)]:
            sub = avg_points[(avg_points['continent']==cont) & (avg_points['year']==year)]
            for _, row in sub.iterrows():
                ax.scatter(row[x], row[y],
                           s=200,
                           c=color,
                           alpha=alpha_val,
                           edgecolor='k',
                           label=cont)
    
    # 参考线显示平均值
    mean_x = height_sel[x].mean()
    mean_y = height_sel[y].mean()
    ax.axvline(mean_x, color='k', linestyle='--')
    ax.text(mean_x, ax.get_ylim()[1]*0.95, f'{x} avg={mean_x:.2f}', ha='center', va='top', fontsize=12)
    ax.axhline(mean_y, color='k', linestyle='--')
    ax.text(ax.get_xlim()[1]*0.98, mean_y, f'{y} avg={mean_y:.2f}', ha='right', va='center', fontsize=12)
    
    ax.set_xlabel(x, fontsize=14)
    ax.set_ylabel(y, fontsize=14)
    ax.set_title(title, fontsize=18)

# Legend 去重
handles, labels = axes[0].get_legend_handles_labels()
by_label = dict(zip(labels, handles))
axes[0].legend(by_label.values(), by_label.keys(), loc='best')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 距离 R 0-20 km
R = np.linspace(0,20,200)

# 定义几组典型参数
params_list = [
    {'A':10, 'B':0.05, 'C':0, 'label':'High A, low B/C'},
    {'A':3,  'B':0.05, 'C':0, 'label':'Low A, low B/C'},
    {'A':7,  'B':0.2,  'C':0, 'label':'Medium A, high B, C≈0'},
    {'A':7,  'B':0.05, 'C':-0.005, 'label':'Medium A, low B, negative C'}
]

plt.figure(figsize=(8,6))

for p in params_list:
    H = p['A'] * np.exp(p['B']*R + p['C']*R**2)
    plt.plot(R,H,label=p['label'])

plt.xlabel('Distance from city center R (km)', fontsize=14)
plt.ylabel('Height H(R)', fontsize=14)
plt.title('Typical city height profiles H(R) = A exp(B*R + C*R^2)', fontsize=16)
plt.legend()
plt.grid(True)
plt.show()
```
